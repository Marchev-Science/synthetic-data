# Chapter 3.1 — Monte Carlo Simulation: Dice Probability Distribution

Monte Carlo simulation estimates quantities that are difficult to compute analytically by generating many random outcomes and accumulating the results. This notebook uses the familiar dice problem to build intuition: what is the probability distribution of the sum when rolling N dice with different numbers of sides?

The analytic solution exists here (combinatorics), but the Monte Carlo approach generalises to arbitrarily complex systems where closed-form solutions do not.

**When to use Monte Carlo simulation:**
- Marginal distributions for all variables are known or can be estimated.
- The system is too complex for an analytic solution.
- You need an estimate of a quantity's distribution (not just its expected value).

> **Interactive:** This notebook prompts for the number of dice and sides. Run it in Jupyter to interact, or replace the `input()` calls with hardcoded values.

## User Input — Number and Type of Dice

```python
%reset -f
import matplotlib.pyplot as plt
import numpy as np
```

## Probability Computation

We enumerate all possible outcome combinations exhaustively (exact calculation). For large numbers of dice this would be replaced by Monte Carlo sampling.

```python
# Get user input for number of dice and number of sides
n_dice = int(input("Enter the number of dice: "))
sides = []
for i in range(n_dice):
    sides.append(int(input(f"Enter the number of sides for die {i+1}: ")))
```

```python
# Calculate the total number of possible outcomes
n_outcomes = np.prod(sides)
```

```python
# Initialize an empty array to store the probabilities
probabilities = np.zeros((np.sum(sides) - n_dice + 1,))
```

## Visualisation — Probability Distribution of the Sum

```python
# Calculate the probability of each possible outcome
dice_values = [0] * n_dice
for i in range(n_outcomes):
    outcome = sum(dice_values) + n_dice  # Add n_dice to get the outcome value (between n_dice and n_dice * sum(sides))
    probabilities[outcome - n_dice] += 1 / n_outcomes  # Increment the probability of the corresponding outcome
    # Increment dice values for next iteration
    for j in range(n_dice):
        dice_values[j] += 1
        if dice_values[j] < sides[j]:
            break
        dice_values[j] = 0
```

## Summary

The distribution of the sum of dice outcomes demonstrates the Central Limit Theorem in action: as the number of dice increases, the sum distribution approaches a normal distribution regardless of the shape of individual die distributions.
